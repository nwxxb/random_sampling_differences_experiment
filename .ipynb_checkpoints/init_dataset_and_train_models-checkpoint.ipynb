# Download Dataset

```python
import gdown
import os

class DriveContent:
  def __init__(self, file_name, url, target_dir="./data"):
    self.target_dir = target_dir
    self.file_path = os.path.join(target_dir, file_name)
    self.url = url
    self.download_file()

  def download_file(self):
    os.makedirs(self.target_dir, exist_ok=True)
    gdown.download(url=self.url, output=self.file_path, quiet=False)
```

# Creating Pipelines and helper and other utilities

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, FunctionTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from imblearn.over_sampling  import RandomOverSampler
from imblearn.under_sampling  import RandomUnderSampler
import nltk
nltk_dir = os.path.abspath("./nltk_data")
os.makedirs(nltk_dir, exist_ok=True)
nltk.download("stopwords", download_dir=nltk_dir)
nltk.data.path.append(nltk_dir)
from nltk.tokenize import word_tokenize
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from nltk.corpus import stopwords
from mpstemmer import MPStemmer
from sklearn.base import BaseEstimator, TransformerMixin
```

## create helper/stateless transformer (will be converted to transformer using FunctionTransformer)

```python
def normalize_unicode(X):
  # try to remove 𝗜𝗻𝘀𝘁𝗮𝗴𝗿𝗮𝗺, ˢᵉᵐᵘᵃ, 𝐊𝐢𝐭𝐚, etc..
  X = X.copy()
  X = X.str.normalize('NFKC')
  return X

def remove_number(X):
  X = X.copy()
  X = X.str.replace(r"\d+", '')
  return X

def remove_feature_generated_mess(X):
  X = X.copy()
  X = X.str.replace(r"\[\[\^[(face|space)\-]{0,6}[\da-f]{8}-[\da-f]{4}-[\da-f]{4}-[\da-f]{4}-[\da-f]{12}\]\]", '')
  return X

def remove_underscore(X):
  X = X.copy()
  X = X.str.replace(r"_", ' ')
  return X

def remove_non_alphabetic(X):
  regex = r"[\u0600-\u06ff]|[\ua980-\ua9df]|[\u1b80-\u1bbf]|[\u1cc0-\u1ccf]|[\u1b00-\u1b7f]|[\u1bc0-\u1bff]|[\u4e00-\u9fff]|[\u0590-\u05ff]|[\ufb1d-\ufb4f]"
  X = X.copy()
  X = X.str.replace(regex, '')
  return X

def remove_kata_ulang(X):
  X = X.copy()
  X = X.str.replace(r"([a-z])(2)", r'\1', regex=True)
  return X

def remove_laughing(X): #TODO
  X = X.copy()
  X = X.str.replace(r"\b(?:[akiq]{0,2}[hwx]{1,2}[akiq]{0,2}){2,}h?\b", '', regex=True)
  return X

def test_print(X, payload):
  print(payload)
  return X
```

## loading stop words for tf_idf

```python
# get stop words from Sastrawi
sastrawi_stop_words_factory = StopWordRemoverFactory()
indo_stop_words = sastrawi_stop_words_factory.get_stop_words()
# get english stop words from nltk
english_stop_words = stopwords.words('english')
# additional stop
additional_stop_words = [
    'dengan', 'ia', 'bahwa', 'oleh', 'http', 'https', 'com', 'org', 'www', 'html', 'lo', 'gw', 'loe', 'gue', 'yg',
    'kamu', 'tdk', 'elo', 'aku', 'daku', 'koe', 'apa', 'bagaimana', 'kapan', 'kpn', 'knp', 'apkh', 'siapa', 'merekalah',
    'dgn', 'msh', 'jg', 'lg', 'lgi', 'kok', 'sy', 'nim', 'nims', 'wah', 'tsb', 'bukan', 'bgt', 'banget', 'buanget', 'gimana', 'gmn',
    'anim', 'enggak', 'nih', 'ku', 'jadi', 'iy', 'gk', 'gak', 'ga', 'tapi', 'tp', 'iya', 'aja', 'aj', 'semua', 'kan',
    'rk', 'klo', 'kalo', 'utk', 'mu', 'sih', 'hai', 'ndak', 'si', 'yaelah', 'yakan', 'tidaklah', 'kalau', 'yo', 'tuh', 'yah',
    'yen', 'wis', 'wes', 'walaupun', 'wae', 'tsbt', 'yup', 'yak', 'waduh', 'woalah', 'yuk', 'tu', 'yes', 'wew', 'untk', 'ngga'
]

stop_words = indo_stop_words + english_stop_words + additional_stop_words
```

## Creating Pipeline

```python
def show_params(pipe_obj):
  params = pipe_obj.get_params().copy()
  for k, v in params.items():
    if '__' in k:
      print(f"{k}: {v}")
```

```python
from imblearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB

NB = Pipeline([
 ('normalize_unicode', FunctionTransformer(
     normalize_unicode
 )),
 ('remove_generated_mess', FunctionTransformer(
     remove_feature_generated_mess
 )),
 ('remove_underscore', FunctionTransformer(
     remove_underscore
 )),
 ('remove_number', FunctionTransformer(
     remove_number
 )),
 ('remove_non_alphabetic_char', FunctionTransformer(
     remove_non_alphabetic
 )),
 ('remove_kata_ulang', FunctionTransformer(
     remove_kata_ulang
 )),
 ('remove_laughing', FunctionTransformer(
     remove_laughing
 )),
 ('tf_idf', TfidfVectorizer(
     stop_words=stop_words
 )),
 ('algo', MultinomialNB())
])

# random over sampling
NB_imb_ROS = Pipeline([
 ('normalize_unicode', FunctionTransformer(
     normalize_unicode
 )),
 ('remove_generated_mess', FunctionTransformer(
     remove_feature_generated_mess
 )),
 ('remove_underscore', FunctionTransformer(
     remove_underscore
 )),
 ('remove_number', FunctionTransformer(
     remove_number
 )),
 ('remove_non_alphabetic_char', FunctionTransformer(
     remove_non_alphabetic
 )),
 ('remove_kata_ulang', FunctionTransformer(
     remove_kata_ulang
 )),
 ('remove_laughing', FunctionTransformer(
     remove_laughing
 )),
 ('tf_idf', TfidfVectorizer(
     stop_words=stop_words
 )),
 ('sampler', RandomOverSampler(
     sampling_strategy='not majority',
     random_state=RANDOM_STATE
 )),
 ('algo', MultinomialNB())
])

# random under sampling
NB_imb_RUS = Pipeline([
 ('normalize_unicode', FunctionTransformer(
     normalize_unicode
 )),
 ('remove_generated_mess', FunctionTransformer(
     remove_feature_generated_mess
 )),
 ('remove_underscore', FunctionTransformer(
     remove_underscore
 )),
 ('remove_number', FunctionTransformer(
     remove_number
 )),
 ('remove_non_alphabetic_char', FunctionTransformer(
     remove_non_alphabetic
 )),
 ('remove_kata_ulang', FunctionTransformer(
     remove_kata_ulang
 )),
 ('remove_laughing', FunctionTransformer(
     remove_laughing
 )),
 ('tf_idf', TfidfVectorizer(
     stop_words=stop_words
 )),
 ('sampler', RandomUnderSampler(
     sampling_strategy='not minority',
     random_state=RANDOM_STATE
 )),
 ('algo', MultinomialNB())
])
```

## Cross Validation to get best parameter

```python
from sklearn.model_selection import GridSearchCV
```

```python
from multiprocessing import cpu_count
n_cpus = cpu_count() # 8
n_cpus # 8
```

```python
# I decide to use 4 n_jobs here, the value heavily depends on your own pc's specs
n_jobs = 4
```

```python
from datetime import datetime as dt
from datetime import timezone

pipes = [
    NB,
    NB_imb_RUS,
    NB_imb_ROS
]
for pipe in pipes:
    start = dt.now(timezone.utc)
    NB_parameters = {
        'tf_idf__use_idf': [True, False],
        'tf_idf__ngram_range': [(1, 1), (1, 2), (1,3)],
        'algo__alpha': [0.00001, 0.0001, 0.001, 0.1, 1, 10, 100,1000],
        'algo__fit_prior': [True, False]
    }
    
    NB_search = GridSearchCV(
        pipe, param_grid=NB_parameters,
        cv=4, verbose=1, scoring='f1_macro',
        n_jobs=n_jobs
    )
    
    NB_search.fit(X_train, y_train)
    fit_seconds = (dt.now(timezone.utc) - start).total_seconds()
    
    pipe_steps_dict = {name: type(step).__name__ for name, step in NB_search.best_estimator_.steps}

    experiment = {
        "meta": {
            "algorithm": pipe_steps_dict['algo'],
            "sampler": pipe_steps_dict.get('sampler'),
            "scoring": NB_search.scoring,
            "random_state": RANDOM_STATE,
            "fit_seconds": fit_seconds,
            "created_at": dt.now(timezone.utc).isoformat()
        },
        'search_result': NB_search,
    }

    import pickle
    
    file_name = "_".join(str(x) for x in [
        start.strftime("%Y%m%d_%H%M%S"),
        pipe_steps_dict['algo'],
        pipe_steps_dict.get('sampler'),
        NB_search.scoring
    ] if x is not None)
    
    os.makedirs("./pickled_models", exist_ok=True)
    pickle.dump(experiment, open(f"./pickled_models/{file_name}.pkl", 'wb'))
    
    print(NB_search.score(X_train, y_train), NB_search.best_score_, NB_search.score(X_test, y_test))
```

```python
def normalize_unicode(X):
  # try to remove 𝗜𝗻𝘀𝘁𝗮𝗴𝗿𝗮𝗺, ˢᵉᵐᵘᵃ, 𝐊𝐢𝐭𝐚, etc..
  X = X.copy()
  X = X.str.normalize('NFKC')
  return X

def remove_number(X):
  X = X.copy()
  X = X.str.replace(r"\d+", '')
  return X

def remove_feature_generated_mess(X):
  X = X.copy()
  X = X.str.replace(r"\[\[\^[(face|space)\-]{0,6}[\da-f]{8}-[\da-f]{4}-[\da-f]{4}-[\da-f]{4}-[\da-f]{12}\]\]", '')
  return X

def remove_underscore(X):
  X = X.copy()
  X = X.str.replace(r"_", ' ')
  return X

def remove_non_alphabetic(X):
  regex = r"[\u0600-\u06ff]|[\ua980-\ua9df]|[\u1b80-\u1bbf]|[\u1cc0-\u1ccf]|[\u1b00-\u1b7f]|[\u1bc0-\u1bff]|[\u4e00-\u9fff]|[\u0590-\u05ff]|[\ufb1d-\ufb4f]"
  X = X.copy()
  X = X.str.replace(regex, '')
  return X

def remove_kata_ulang(X):
  X = X.copy()
  X = X.str.replace(r"([a-z])(2)", r'\1', regex=True)
  return X

def remove_laughing(X): #TODO
  X = X.copy()
  X = X.str.replace(r"\b(?:[akiq]{0,2}[hwx]{1,2}[akiq]{0,2}){2,}h?\b", '', regex=True)
  return X

def stem_indo(X):
  return X

def test_print(X, payload):
  print(payload)
  return X
```
